```python
import pandas as pd
import numpy as np
import sklearn as skl
import feedparser
from deep_translator import MyMemoryTranslator
```

```python
links= {
    1:"https://www.aitrends.com/feed/",
    2:"https://www.techrepublic.com/rssfeeds/topic/cybersecurity/",
    3: "https://machinelearningmastery.com/blog/feed",
    4: "https://feeds.feedburner.com/gov/gnCH",
    5: "https://www.nexojornal.com.br/rss.xml",
    6: "https://guiadoestudante.abril.com.br/feed/"

}

def alimentador_noticias(links):
  
    noticias = []

    # Coleta as notícias de todos os feeds
    for chave, link in links.items():

        print(chave, "-", link)

        feed = feedparser.parse(link)

        for entry in feed.entries:

            noticias.append({
                "title": entry.get("title", ""),
                "link": entry.get("link", ""),
                "published": entry.get("published", ""),
                "summary": entry.get("summary", "")
            })


    return pd.DataFrame(noticias)
```

```python
dados_jornal = alimentador_noticias(links)
```

```python
dados_jornal["published"] = pd.to_datetime(dados_jornal["published"], format="mixed", utc=True)
dados_jornal["published"] = dados_jornal["published"].dt.tz_convert("America/Sao_Paulo")
dados_jornal["published"] = dados_jornal["published"].dt.strftime("%d/%m/%Y %H:%M:%S")
dados_jornal.head()
```

```python
def traduzir(texto):
    if not texto:
        return ""

    return MyMemoryTranslator(
        source="english",
        target="portuguese",
    ).translate(texto)

dados_jornal["title"]  = dados_jornal["title"].apply(traduzir)
```

```python
html = """
<!DOCTYPE html>

<html lang="pt-br">

<head>

    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <link rel="stylesheet" href="static/style.css">
    <title> Jornal </title>

</head>


<body>


<header class="cabecalho">

    <div class="linha-topo"></div>

    <h1> NOTÍCIAS </h1>

    <p> Catálogo de notícias </p>

</header>



<main>

"""
if len(dados_jornal) > 0:

    primeira = dados_jornal.iloc[0]

    html += f"""


    <section class="outras">

        <div class="titulo-secao">

            <span>NOTÍCIAS</span>

        </div>


        <div class="grid-noticias">

    """


    for _, linha in dados_jornal.iloc[1:7].iterrows():

        html += f"""

            <article class="card">

                <span class="etiqueta">
                    NOTÍCIA
                </span>

                <h3>
                    {linha['title']}
                </h3>

                <p class="data">
                    {linha['published']}
                </p>

                <p class="resumo">
                    {linha['summary']}
                </p>

                <a
                    href="{linha['link']}"
                    target="_blank"
                    class="ler-mais"
                >
                    Ler mais →
                </a>

            </article>

        """


    html += """

        </div>

    </section>

    """

html += """

</main>


<footer>

    <div class="footer-conteudo">

        <h2> JORNAL</h2>

        <p>
            Catálogo de notícias
        </p>

    </div>

</footer>


</body>

</html>

"""
```

```python
with open("jornal.html", "w", encoding="utf-8") as arquivo:

    arquivo.write(html)


print("Jornal criado com sucesso!")
```
