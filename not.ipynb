```python
import numpy as np
import pandas as pd
import matplotlib as plt
import seaborn as sns
import seaborn as sns
from io import StringIO
import feedparser
```

```python
links= {
    1:"https://www.aitrends.com/feed/",
    2:"https://www.techrepublic.com/rssfeeds/topic/cybersecurity/",
    3: "https://machinelearningmastery.com/blog/feed",
    4: "https://feeds.feedburner.com/gov/gnCH",
    5: "https://www.nexojornal.com.br/rss.xml",
    6: "https://guiadoestudante.abril.com.br/feed/"

}

def alimentador_noticias(links):
  #Faz a coleta dos dados nas páginas xml
  for chave, link in links.items():
    print(chave,' - ',link)
    feed = feedparser.parse(link)
    print(feed)
  # Acessa os dados coletados e manipula para deixar no formato para uso
  for entry in feed.entries:
      noticia_limpa = pd.DataFrame(feed.entries)
      print("Título:",entry.title)
      noticia = str.replace(entry.description,".",".\n")
      noticia = str.replace(noticia,";",";\n")
      noticia = str.replace(noticia," • "," • \n")
      print("Notícia:",noticia)


  return noticia_limpa

noticia = alimentador_noticias(links)
```
